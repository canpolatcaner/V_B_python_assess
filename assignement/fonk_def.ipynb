```python
sayi1 = input("Lütfen işlem yaptırmak istediğiniz bir sayı giriniz\t:")
sayi2 = input("Lütfen işlem yaptırmak istediğiniz diğer sayıyı giriniz\t:")
def islem(): 
    print(f"Seçtiğiniz 1'inci sayı: {sayi1}\n Seçtiğiniz 2'nci sayı: {sayi2}.\n")
    secim = input("Lütfen yapmak istediğiniz işlemi seçiniz\t:\n '+''-''/''*'")
    if secim == "+":
        print(f"Toplama işlemini seçtiniz.\n {sayi1} + {sayi2} = {sayi1+sayi2}")
islem()
    
          
```
